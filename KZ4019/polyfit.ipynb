```python
import numpy as np
import pandas as pd

# Given points (x -> y)
x = np.array([1.23968, 1.24468, 1.24968, 1.25468, 1.25968, 1.26468, 1.26968])
y = np.array([
    -150.270002732,
    -150.270148465,
    -150.270234419,
    -150.270262653,
    -150.270235161,
    -150.270153867,
    -150.270020632,
])

# Second-order polynomial fit: y = ax^2 + bx + c
coef = np.polyfit(x, y, 2)
poly = np.poly1d(coef)

# Vertex (extremum) of the parabola
x_vertex = -coef[1] / (2 * coef[0])
y_vertex = poly(x_vertex)

# Evaluate fitted curve on a fine grid for plotting or inspection
x_fit = np.linspace(x.min(), x.max(), 200)
y_fit = poly(x_fit)

print("Coefficients [a, b, c]:", coef)
print(f"Vertex at x = {x_vertex:.6f}, y = {y_vertex:.9f}")

# Show the original data vs fitted values at the same x for quick comparison
fit_at_data = poly(x)
df = pd.DataFrame({"x": x, "y_obs": y, "y_fit": fit_at_data})
display(df)
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.plot(x_fit, y_fit, label="2nd-order fit", color="tab:blue")
plt.scatter(x, y, label="data", color="tab:orange", zorder=3)
plt.scatter([x_vertex], [y_vertex], color="red", marker="x", s=60, label="vertex")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Quadratic fit")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
k=2*coef[0]  # k = 2a for a parabola y = ax^2 + bx + c
print (f"Estimated curvature k = {k:.6f}")
```
